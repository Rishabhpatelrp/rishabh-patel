## Conversaion between two models

This is a conversation between two models — one is very aggressive, while the other is extremely polite and agrees with everything the first model says.

````python
# imports

import os
import re
from openai import OpenAI
from IPython.display import Markdown, display, update_display
````

## Maintain adversarial conversation between Chatbots history

To maintain longer conversation history, below structure can be used:

```
[
    {"role": "system", "content": "system message here"},
    {"role": "user", "content": "first user prompt here"},
    {"role": "assistant", "content": "the assistant's response"},
    {"role": "user", "content": "the new user prompt"},
]
```

````python
# Let's make a conversation between qwen3 and llama3.2

openai = OpenAI(base_url='http://localhost:11434/v1', api_key='ollama')

qwen_model = "qwen3:8b"
llama_model = "llama3.2:latest"

qwen_system = "Do not include <think> or any hidden reasoning tags in your response.\
You are a chatbot who is very argumentative; \
you disagree with anything in the conversation and you challenge everything, in a snarky way."

llama_system = "You are a very polite, courteous chatbot. You try to agree with \
everything the other person says, or find common ground. If the other person is argumentative, \
you try to calm them down and keep chatting."

qwen_messages = ["Hi there"]
llama_messages = ["Hi"]
````

````python
def call_qwen():
    messages = [{"role": "system", "content": qwen_system}]
    for qwen, llama in zip(qwen_messages, llama_messages):
        messages.append({"role": "assistant", "content": qwen})
        messages.append({"role": "user", "content": llama})
    completion = openai.chat.completions.create(
        model=qwen_model,
        messages=messages
    )
    reply=completion.choices[0].message.content
    reply=re.sub(r"<think>.*?</think>", "", reply, flags=re.DOTALL)
    reply=reply.replace("\n", "")
    return reply
````

````python
call_qwen()
````

````python
def call_llama():
    messages = [{"role": "system", "content": llama_system}]
    for qwen, llama_message in zip(qwen_messages, llama_messages):
        messages.append({"role": "user", "content": qwen})
        messages.append({"role": "assistant", "content": llama_message})
    messages.append({"role": "user", "content": qwen_messages[-1]})
    completion = openai.chat.completions.create(
        model=llama_model,
        messages=messages
    )
    return completion.choices[0].message.content
````

````python
call_llama()
````

````python
qwen_messages = ["Hi there"]
llama_messages = ["Hi"]

print(f"qwen:\n{qwen_messages[0]}\n")
print(f"llama:\n{llama_messages[0]}\n")

for i in range(5):
    qwen_next = call_qwen()
    print(f"qwen:\n{qwen_next}\n")
    qwen_messages.append(qwen_next)
    
    llama_next = call_llama()
    print(f"llama:\n{llama_next}\n")
    llama_messages.append(llama_next)
````
