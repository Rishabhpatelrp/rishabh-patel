# Explain code

Build a tool that takes a technical question and responds with an explanation.

````python
# imports
import os
from dotenv import load_dotenv
from IPython.display import Markdown, display, update_display
from openai import OpenAI
````

````python
# constants

MODEL_GPT = 'gpt-4o-mini'
MODEL_LLAMA = 'llama3.2'
````

````python
# set up environment
# Consider creating ENV file before executing it, not required if running model locally
load_dotenv(override=True)
api_key = os.getenv('OPENAI_API_KEY')

if api_key and api_key.startswith('sk-proj-') and len(api_key)>10:
    print("API key looks good so far")
else:
    print("There might be a problem with your API key? Please visit the troubleshooting notebook!")
````

````python
# here is the question; type over this to ask something new
system_prompt = "Please explain what this code does and why"

def get_user_prompt(code):
    return f"""code is
    {code}"""

# Please explain what this code does and why:
# yield from {book.get("author") for book in books if book.get("author")}
````

````python
# Get gpt-4o-mini to answer, with streaming
openai = OpenAI(api_key=api_key)
MODEL = MODEL_GPT
````

````python
# Get Llama 3.2 to answer with stream
openai = OpenAI(base_url='http://localhost:11434/v1', api_key='ollama')
MODEL = MODEL_LLAMA
````

````python
def get_code_explaination(code):
    messages=[
            {"role": "system", "content": system_prompt},
            {"role": "user", "content": get_user_prompt(code)}
    ];
    stream = openai.chat.completions.create(
        model=MODEL, 
        messages=messages,
        stream=True
    )
    
    response = ""
    display_handle = display(Markdown(""), display_id=True)
    for chunk in stream:
        response += chunk.choices[0].delta.content or ''
        response = response.replace("```","").replace("markdown", "")
        update_display(Markdown(response), display_id=display_handle.display_id)
````

````python
code = 'yield from {book.get("author") for book in books if book.get("author")}'
get_code_explaination(code)
````
